**Exemplo - Motor de Indução**

```python
# Inserção de dados do motor pelo usuário
# Potência útil do motor (PU)
PU = int(input("Digite o valor da potencia util (em wats): "))
# Rendimento do motor (n)
n = float(input("Digite o valor do rendimento (de 0 a 1): "))
```

```python
print(PU, n)
```

```python
# Calculo da potência ativa (P)
# n = PU / P
P = PU / n
print(f"A potência ativa do motor será de {int(P)} W") 
```
